# Save a pretrained checkpoint for offline use

The competition scores your notebook with **internet access turned off**, so `from_pretrained("microsoft/deberta-v3-small")` cannot download anything at submission time. This helper downloads a checkpoint once and saves it as notebook output, which you then attach to your project notebook as an input.

1. Create a new Kaggle notebook, paste these cells, and turn **Internet on** (Settings → Internet). No GPU is needed.
2. Set `MODEL_ID`, then **Save Version → Save & Run All (Commit)**.
3. In your project notebook: **Add Input → Your Work → Notebooks**, then select this notebook. Its output appears under `/kaggle/input/`, and the starter notebook finds it automatically.

Repeat with a different `MODEL_ID` if you try another checkpoint. Check each checkpoint's model card for its license.

```python
from pathlib import Path
import os
from huggingface_hub import snapshot_download

MODEL_ID = "microsoft/deberta-v3-small"   # any public Hugging Face checkpoint
root_dir = Path(os.environ.get("PATENT_WEIGHTS_DIR", "/kaggle/working" if Path("/kaggle/working").is_dir() else "patent_weights"))
save_dir = root_dir / MODEL_ID.split("/")[-1]

snapshot_download(
    repo_id=MODEL_ID,
    revision="a36c739020e01763fe789b4b85e2df55d6180012",  # same pretrained revision as the local experiments
    local_dir=save_dir,
    # PyTorch weights, configs, and tokenizer files only (skips TensorFlow/Flax/ONNX copies)
    allow_patterns=["*.json", "*.safetensors", "*.bin", "*.model", "*.txt"],
)
print(sorted(p.name for p in save_dir.iterdir()))
```
