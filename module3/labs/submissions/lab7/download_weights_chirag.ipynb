Run once on Kaggle with internet on. Save a notebook version, then attach its output to the patent notebook and switch that notebook's internet off. These are public pretrained checkpoints, not competition solutions. MiniLM is loaded with mean pooling and L2 normalization, as in its model card.

```python
from pathlib import Path
from transformers import AutoModel, AutoModelForSequenceClassification, AutoTokenizer

root = Path('/kaggle/working') if Path('/kaggle/working').is_dir() else Path('patent_weights')
for model_id, cls in [('microsoft/deberta-v3-small',AutoModelForSequenceClassification),
                       ('sentence-transformers/all-MiniLM-L6-v2',AutoModel)]:
    dest = root / model_id.rsplit('/',1)[-1]
    tok = AutoTokenizer.from_pretrained(model_id)
    kwargs = {'num_labels':1} if cls is AutoModelForSequenceClassification else {}
    model = cls.from_pretrained(model_id,**kwargs)
    tok.save_pretrained(dest)
    model.save_pretrained(dest)
    print('Saved',dest)
    del model
```
